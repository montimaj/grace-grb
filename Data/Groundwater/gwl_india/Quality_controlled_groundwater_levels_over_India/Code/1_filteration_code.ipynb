## Criterion 3: Remove outlier using 3sigma condition for each well. 
## We are not removing any wells; instead, we are removing outliers within each well, resulting in the same number of wells being retained.
## Output from this code includes  one csv, one excel, and folder with timeseries plot for five random wells.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import random
import os

# Load the data
input_file = "/your_path_to_Example_data_folder/Example_data/Output/3_India_GWLs_2000_2024_non_negative_obs.csv"
df = pd.read_csv(input_file)

# Separate columns
columns = df.columns.tolist()
info_cols = columns[:20]
time_cols = columns[20:]

# Compute statistics before 2-sigma cleaning
stats_before = df[time_cols].agg(['min', 'max', 'mean', 'count', 'std'], axis=1).T
stats_before.columns = df.index
stats_before = stats_before.T  # to align wells as rows

# 2-sigma outlier removal function
def remove_outliers_3sigma(row):
    values = row[time_cols]
    mean = values.mean()
    std = values.std()
    mask = (values - mean).abs() <= 3 * std
    return values.where(mask)

# Apply outlier removal
df_cleaned = df.copy()
df_cleaned[time_cols] = df.apply(remove_outliers_3sigma, axis=1)

# Compute statistics after 3-sigma cleaning
stats_after = df_cleaned[time_cols].agg(['min', 'max', 'mean', 'count', 'std'], axis=1).T
stats_after.columns = df_cleaned.index
stats_after = stats_after.T

# Save cleaned data
output_csv = "/your_path_to_Example_data_folder/Example_data/Output/4_India_GWLs_2000_2024_after_3sigma.csv"
df_cleaned.to_csv(output_csv, index=False)

# Save statistics to Excel
output_excel ="/your_path_to_Example_data_folder/Example_data/Output/4_India_GWLs_2000_2024_after_3sigma_summary.xlsx" 
with pd.ExcelWriter(output_excel, engine='openpyxl') as writer:
    stats_before.to_excel(writer, sheet_name='Stats_Before_3sigma')
    stats_after.to_excel(writer, sheet_name='Stats_After_3sigma')

# Plotting before and after for 5 random wells
output_plot_dir = "/your_path_to_Example_data_folder/Example_data/Output/Plots_timeseries/"
os.makedirs(output_plot_dir, exist_ok=True)

random_indices = random.sample(list(df.index), 5)
for idx in random_indices:
    well_id = df.loc[idx, info_cols[0]]  # assuming first info_col is unique well ID
    ts_before = df.loc[idx, time_cols]
    ts_after = df_cleaned.loc[idx, time_cols]

    plt.figure(figsize=(12, 4))
    plt.plot(time_cols, ts_before, label='Before 3σ', marker='o')
    plt.plot(time_cols, ts_after, label='After 3σ', marker='x')
    plt.title(f"Well: {well_id}")
    plt.xticks(rotation=90)
    plt.xlabel("Time")
    plt.ylabel("GWL")
    plt.legend()
    plt.tight_layout()
    plt.savefig(os.path.join(output_plot_dir, f"{well_id}_3sigma_plot.png"))
    plt.close()

print(f"\n✅ Cleaned data saved to: {output_csv}")
print(f"✅ Statistics saved to: {output_excel}")
print(f"✅ Plots saved to: {output_plot_dir}")
```

## Criterion 5: Remove wells where GWLs repeat consecutively more than twice

This Python program processes groundwater level data from the Central Ground Water Board (CGWB) of India. The goal is to clean and filter the dataset by applying two quality-control conditions and generate cleaned outputs for different time ranges. All output files are saved in a designated folder: /mnt/Data/5hk/Data/sy_data_paper/Data_paper_revision/Processed_atleast_2_in_a_year_2000_2022/, which is automatically created if it doesn’t already exist. For each year range (e.g., 2000–2024 down to 2004–2015), two CSV files are saved: one after applying the first condition (requiring at least two valid groundwater level values among Jan, May, Aug, and Nov per year), and another after applying the second condition (removing wells with values repeating more than twice consecutively). These files are named using the format CGWB_<start_year>_<end_year>_cond1.csv and CGWB_<start_year>_<end_year>_cond2_continuous_repeats.csv. Additionally, a summary Excel file named CGWB_Filtering_Summary.xlsx is saved in the same folder, documenting the number of wells retained after each filtering step across all processed time ranges.

```python
import pandas as pd
import os

# === Load the CGWB data === #
CGWB = pd.read_csv('/your_path_to_Example_data_folder/Example_data/Output/4_India_GWLs_2000_2024_after_3sigma.csv')



# === Metadata columns to retain === #
metadata_columns = ['Station Code', 'Station Name', 'Station Type', 'Agency Name', 'State', 'District', 'Tehsil',
                    'Data Acquisition', 'Block', 'Village', 'Latitude', 'Longitude', 'Data Available From',
                    'Latest  Data Available', 'Type of Well', 'Aquifer Type', 'Well Depth', 'Data Type Code',
                    'Data Type Description', 'Unit']

# === Filtering Functions === #

# Condition 1: At least 2 non-NaNs among Jan, May, Aug, Nov for each year
def has_at_least_2_non_nan(row, year_short):
    columns_for_year = [f'Jan-{year_short}', f'May-{year_short}', f'Aug-{year_short}', f'Nov-{year_short}']
    non_nan_count = sum(pd.notna(row.get(col, None)) for col in columns_for_year)
    return non_nan_count >= 2

# New Condition: Remove rows where any value repeats more than twice continuously
def filter_continuous_repeats(df):
    mask = pd.Series(True, index=df.index)
    for idx, row in df.iterrows():
        # Extract time series values (after metadata columns)
        values = row[20:].values
        # Convert to numeric and handle NaNs
        numeric_values = pd.to_numeric(values, errors='coerce')

        # Check for 3 or more continuous repeats
        count = 1
        prev_val = None
        for val in numeric_values:
            if pd.isna(val):
                count = 1
                prev_val = None
                continue
            if val == prev_val:
                count += 1
                if count > 2:
                    mask[idx] = False
                    break
            else:
                count = 1
                prev_val = val
    return df.loc[mask].reset_index(drop=True)

# === Helper function to generate year suffixes === #
def get_years_short(start_year, end_year):
    return [f"{y:02d}" for y in range(start_year - 2000, end_year - 1999)]

# === Output Directory === #
output_dir = "/your_path_to_Example_data_folder/Example_data/Output/Filtered_GWLs_2000_2024/"
os.makedirs(output_dir, exist_ok=True)

# === Store Summary Statistics === #
summary_counts = []

# === Function to process a given year range === #
def process_year_range(start_year, end_year):
    print(f"Processing: {start_year} to {end_year}")
    year_range = f"{start_year}-{end_year}"
    years_short = get_years_short(start_year, end_year)

    # Condition 1
    filtered_df = CGWB.copy()
    for year_short in years_short:
        filtered_df = filtered_df[filtered_df.apply(lambda row: has_at_least_2_non_nan(row, year_short), axis=1)]

    time_columns = [f'{month}-{year_short}' for year_short in years_short for month in ['Jan', 'May', 'Aug', 'Nov']]
    final_columns = metadata_columns + time_columns
    final_df1 = filtered_df[final_columns]

    # Apply new condition for continuous repeats > 2
    final_df2 = filter_continuous_repeats(final_df1)
    final_df2.to_csv(os.path.join(output_dir, f'CGWB_filtered_wells_{start_year}_{end_year}.csv'), index=False)


# === Process ranges: 2000–2024 to 2000–2015 === #
for end_year in range(2024, 2015 - 1, -1):
    process_year_range(2000, end_year)


print("✅ Processing complete. All files and summary saved.")
```
